```python
import pandas as pd

query = """
SELECT
    MATCH_ID,
    SEASON_ID,
    TEAM1,
    TEAM2,
    CITY,
    VENUE,
    TOSS_WINNER,
    TOSS_DECISION,
    MATCH_WINNER,
    RESULT,
    MATCH_TYPE,
    STAGE
FROM NLP2SQL."test schema".IPLMATCHS
"""

cursor.execute(query)

rows = cursor.fetchall()
columns = [col[0] for col in cursor.description]

df = pd.DataFrame(rows, columns=columns)

print("✅ IPL data loaded successfully!")
print("Total rows:", len(df))
print("Columns:", df.columns.tolist())

df.head(10)
```

```python
print("Shape:", df.shape)

print("\nFirst 10 rows:")
display(df.head(10))

print("\nColumn information:")
print(df.info())

print("\nMissing values:")
print(df.isnull().sum())

print("\nMatch winners:")
print(df["MATCH_WINNER"].value_counts())
```

```python
# Remove rows where important ML fields are missing

df = df.dropna(
    subset=[
        "TEAM1",
        "TEAM2",
        "VENUE",
        "TOSS_WINNER",
        "TOSS_DECISION",
        "MATCH_WINNER"
    ]
).copy()

print("✅ Data cleaned successfully!")
print("Remaining rows:", len(df))
print("Missing values:")
print(df[
    [
        "TEAM1",
        "TEAM2",
        "VENUE",
        "TOSS_WINNER",
        "TOSS_DECISION",
        "MATCH_WINNER"
    ]
].isnull().sum())
```

```python
features = [
    "TEAM1",
    "TEAM2",
    "VENUE",
    "TOSS_WINNER",
    "TOSS_DECISION"
]

X = df[features].copy()
y = df["MATCH_WINNER"].copy()

print("✅ Features and target prepared!")

print("\nFeatures:")
print(X.head())

print("\nTarget:")
print(y.head())
```

```python
from sklearn.preprocessing import LabelEncoder

# Store an encoder for each feature
encoders = {}

for column in features:
    encoder = LabelEncoder()
    X[column] = encoder.fit_transform(X[column].astype(str))
    encoders[column] = encoder

# Encode the target (MATCH_WINNER)
target_encoder = LabelEncoder()
y = target_encoder.fit_transform(y.astype(str))

print("✅ Encoding completed!")

print("\nEncoded features:")
print(X.head())

print("\nEncoded target:")
print(y[:10])

print("\nTeams/classes:")
print(list(target_encoder.classes_))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("✅ Data split completed!")

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Training target:", len(y_train))
print("Testing target:", len(y_test))
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    class_weight="balanced"
)

model.fit(X_train, y_train)

print("✅ Random Forest model trained successfully!")
```

```python
from sklearn.metrics import accuracy_score

# Make predictions on test data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("✅ Prediction completed!")
print("Model Accuracy:", round(accuracy * 100, 2), "%")

```

```python
from sklearn.metrics import classification_report

print("Classification Report:\n")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=target_encoder.classes_,
        zero_division=0
    )
)
```

```python
import joblib

joblib.dump(model, "ipl_prediction_model.pkl")
joblib.dump(encoders, "ipl_feature_encoders.pkl")
joblib.dump(target_encoder, "ipl_target_encoder.pkl")

print("✅ Model files saved successfully!")
```

```python
def predict_winner(team1, team2, venue, toss_winner, toss_decision):

    input_data = pd.DataFrame({
        "TEAM1": [team1],
        "TEAM2": [team2],
        "VENUE": [venue],
        "TOSS_WINNER": [toss_winner],
        "TOSS_DECISION": [toss_decision]
    })

    # Encode input values
    for column in features:
        input_data[column] = encoders[column].transform(
            input_data[column].astype(str)
        )

    # Predict winner
    prediction = model.predict(input_data)[0]

    winner = target_encoder.inverse_transform([prediction])[0]

    # Prediction probability
    probabilities = model.predict_proba(input_data)[0]
    confidence = max(probabilities)

    return winner, confidence


print("✅ Prediction function created successfully!")
```

```python
print("Teams:")
print(list(encoders["TEAM1"].classes_))

print("\nVenues:")
print(list(encoders["VENUE"].classes_))

print("\nToss decisions:")
print(list(encoders["TOSS_DECISION"].classes_))
```

```python
team1 = "Mumbai Indians"
team2 = "Chennai Super Kings"
venue = "Wankhede Stadium"
toss_winner = "Mumbai Indians"
toss_decision = "bat"

winner, confidence = predict_winner(
    team1,
    team2,
    venue,
    toss_winner,
    toss_decision
)

print("🏏 IPL MATCH PREDICTION")
print("-----------------------------")
print("Team 1       :", team1)
print("Team 2       :", team2)
print("Venue        :", venue)
print("Toss Winner  :", toss_winner)
print("Toss Decision:", toss_decision)
print("-----------------------------")
print("Predicted Winner:", winner)
print("Confidence      :", f"{confidence * 100:.2f}%")
```

```python
def ipl_prediction_agent(
    team1,
    team2,
    venue,
    toss_winner,
    toss_decision
):
    try:
        winner, confidence = predict_winner(
            team1,
            team2,
            venue,
            toss_winner,
            toss_decision
        )

        return {
            "Team 1": team1,
            "Team 2": team2,
            "Venue": venue,
            "Toss Winner": toss_winner,
            "Toss Decision": toss_decision,
            "Predicted Winner": winner,
            "Confidence": f"{confidence * 100:.2f}%"
        }

    except Exception as e:
        return {
            "error": str(e)
        }


print("✅ IPL Prediction Agent created successfully!")
```

```python
import gradio as gr

# Get available values from the encoders
teams = list(encoders["TEAM1"].classes_)
venues = list(encoders["VENUE"].classes_)
toss_decisions = list(encoders["TOSS_DECISION"].classes_)


def predict_from_ui(team1, team2, venue, toss_winner, toss_decision):

    # Prevent selecting the same team
    if team1 == team2:
        return "⚠️ Team 1 and Team 2 must be different."

    result = ipl_prediction_agent(
        team1,
        team2,
        venue,
        toss_winner,
        toss_decision
    )

    if "error" in result:
        return f"❌ Error: {result['error']}"

    return f"""
🏏 IPL MATCH PREDICTION
━━━━━━━━━━━━━━━━━━━━━━━━

🏏 Team 1: {result['Team 1']}
🏏 Team 2: {result['Team 2']}

📍 Venue: {result['Venue']}

🪙 Toss Winner: {result['Toss Winner']}
🪙 Toss Decision: {result['Toss Decision']}

━━━━━━━━━━━━━━━━━━━━━━━━

🏆 Predicted Winner:
### {result['Predicted Winner']}

📊 Model Confidence:
### {result['Confidence']}
"""


with gr.Blocks(title="IPL Match Winner Predictor") as demo:

    gr.Markdown(
        """
        # 🏏 IPL Match Winner Predictor

        ### 🤖 Machine Learning Prediction System

        Enter the match details below to predict the winner.
        """
    )

    with gr.Row():

        team1 = gr.Dropdown(
            choices=teams,
            label="🏏 Team 1",
            value="Mumbai Indians"
        )

        team2 = gr.Dropdown(
            choices=teams,
            label="🏏 Team 2",
            value="Chennai Super Kings"
        )

    venue = gr.Dropdown(
        choices=venues,
        label="📍 Venue",
        value="Wankhede Stadium"
    )

    toss_winner = gr.Dropdown(
        choices=teams,
        label="🪙 Toss Winner",
        value="Mumbai Indians"
    )

    toss_decision = gr.Dropdown(
        choices=toss_decisions,
        label="🪙 Toss Decision",
        value="bat"
    )

    predict_button = gr.Button(
        "🔮 Predict Match Winner",
        variant="primary"
    )

    output = gr.Markdown(
        label="Prediction Result"
    )

    predict_button.click(
        fn=predict_from_ui,
        inputs=[
            team1,
            team2,
            venue,
            toss_winner,
            toss_decision
        ],
        outputs=output
    )


demo.launch(share=True)
```

```python
def predict_from_ui(team1, team2, venue, toss_winner, toss_decision):

    if team1 == team2:
        return "⚠️ Team 1 and Team 2 must be different."

    try:
        input_data = pd.DataFrame({
            "TEAM1": [team1],
            "TEAM2": [team2],
            "VENUE": [venue],
            "TOSS_WINNER": [toss_winner],
            "TOSS_DECISION": [toss_decision]
        })

        # Encode input values
        for column in features:
            input_data[column] = encoders[column].transform(
                input_data[column].astype(str)
            )

        # Predict
        prediction = model.predict(input_data)[0]
        winner = target_encoder.inverse_transform([prediction])[0]

        # Get probabilities
        probabilities = model.predict_proba(input_data)[0]

        classes = list(target_encoder.classes_)

        team1_index = classes.index(team1)
        team2_index = classes.index(team2)

        team1_probability = probabilities[team1_index] * 100
        team2_probability = probabilities[team2_index] * 100

        return f"""
# 🏏 IPL MATCH PREDICTION

### {team1} 🆚 {team2}

**📍 Venue:** {venue}

**🪙 Toss Winner:** {toss_winner}

**🪙 Toss Decision:** {toss_decision}

---

## 🏆 Predicted Winner

# {winner}

### 📊 Prediction Probabilities

**{team1}:** {team1_probability:.2f}%

**{team2}:** {team2_probability:.2f}%

---

⚠️ *This is a machine-learning prediction and is not a guarantee of the actual match result.*
"""

    except Exception as e:
        return f"❌ Error: {str(e)}"


print("✅ Advanced prediction function created!")
```
